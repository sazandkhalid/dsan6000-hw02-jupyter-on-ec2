```python
import boto3
import os

s3 = boto3.client('s3')
region = 'us-east-1'
netid = 'sk2153' 
bucket_name = f'dsan6000-{netid}'
local_dir = 'data'
s3_prefix = 'wikipedia-hourly/'
```

```python
if region == 'us-east-1':
    s3.create_bucket(Bucket=bucket_name)
else:
    s3.create_bucket(
        Bucket=bucket_name,
        CreateBucketConfiguration={'LocationConstraint': region}
    )
print(f"Created bucket: {bucket_name}")
```

```python
files = [f for f in os.listdir(local_dir) if f.endswith('.parquet')]

for filename in files:
    local_path = os.path.join(local_dir, filename)
    s3_key = s3_prefix + filename
    s3.upload_file(local_path, bucket_name, s3_key)
    print(f"Uploaded {filename}")

print(f"Total uploaded: {len(files)}")
```

```python
response = s3.list_objects_v2(Bucket=bucket_name, Prefix=s3_prefix)
uploaded = [obj['Key'] for obj in response.get('Contents', [])]
print(f"Files in bucket: {len(uploaded)}")
for key in uploaded:
    print(key)
```
